### 실습 3. Dictionary로 단어 빈도 원리 이해하기

 

빈도 분석의 핵심은 매우 단순합니다.

 

같은 단어가 등장할 때마다 해당 단어의 숫자를 1씩 증가시키면 됩니다.

 

예를 들어 다음 단어가 있다고 가정합니다.

```python
sample_words = [
    "데이터",
    "분석",
    "파이썬",
    "데이터",
    "AI",
    "분석",
    "데이터",
]

# 방법 1: if문으로 직접 확인해서 세기
word_counts = {}
for word in sample_words:
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

print("방법 1 결과:", word_counts)

# 방법 2: get()을 이용해 더 간단하게 세기
word_counts2 = {}
for word in sample_words:
    word_counts2[word] = word_counts2.get(word, 0) + 1

print("방법 2 결과:", word_counts2)
```

## 실행 결과 확인

`if`문 방식과 `get()` 방식으로 단어 빈도를 각각 세어본 결과, 두 방법 모두 동일하게 나왔다.

- 데이터: 3, 분석: 2, 파이썬: 1, AI: 1
- `if word in word_counts`로 직접 확인하는 방식과 `word_counts.get(word, 0)`으로 간단히 세는 방식은 결과가 같은, 동일한 원리의 코드임을 확인

### 실습 4. Counter로 같은 작업 간단하게 처리하기

##### Counter로 빈도 계산하기

`collections.Counter`를 사용하면 단어 빈도를 더 쉽게 계산할 수 있다.

- `Counter(sample_words)` → 각 단어의 등장 횟수를 자동으로 세어줌 (앞서 Dictionary로 직접 만든 것과 결과 동일)
- `.most_common()` → 많이 등장한 순서대로 전체 정렬
- `.most_common(3)` → 상위 3개만 확인

즉, Dictionary는 빈도를 세는 원리를 이해하기 위한 방법이고, Counter는 같은 결과를 더 간단한 코드로 얻는 도구다.

```python
from collections import Counter

# 전체 빈도 계산
sample_counter = Counter(sample_words)
print("전체 빈도:", sample_counter)

# 많이 등장한 순서로 전체 정렬해서 확인
print("\n빈도순 전체 정렬:", sample_counter.most_common())

# 상위 3개만 확인
print("\n상위 3개:", sample_counter.most_common(3))
```

## 실행 결과 확인

`Counter`로 sample_words의 빈도를 계산한 결과, 앞서 Dictionary로 직접 세었던 결과와 동일하게 나왔다.

- 전체 빈도: 데이터 3, 분석 2, 파이썬 1, AI 1
- `most_common()`으로 빈도 높은 순 정렬 확인
- `most_common(3)`으로 상위 3개(데이터, 분석, 파이썬)만 확인

## 실습 6. 정규표현식으로 간단한 토큰화

한글 형태소 분석 전에, 정규표현식으로 단순하게 단어를 나눠보는 단계다.

- `re.findall(r"[가-힣A-Za-z0-9]+", str(text).lower())`로 한글/영어/숫자 덩어리만 추출
- 예: "AI 시대의 데이터 분석, Python으로 시작하기!" → `['ai', '시대의', '데이터', '분석', 'python으로', '시작하기']`
- 한계: '시대의', 'python으로', '시작하기'처럼 조사나 어미가 단어에 붙어서 사람이 생각하는 의미 단위와 다르게 잘림
- 이 한계 때문에 다음 단계에서 형태소 분석을 사용함

```python
import re

def simple_tokenize(text):
    # 한글, 영어, 숫자로 이루어진 덩어리만 뽑아서 리스트로 반환
    return re.findall(r"[가-힣A-Za-z0-9]+", str(text).lower())

# 테스트 문장으로 함수 동작 확인
sample_title = "AI 시대의 데이터 분석, Python으로 시작하기!"
print(simple_tokenize(sample_title))
```

## 실행 결과 확인

정규표현식 기반 `simple_tokenize()` 함수를 테스트 문장에 적용한 결과, 예상대로 나왔다.

- 결과: `['ai', '시대의', '데이터', '분석', 'python으로', '시작하기']`
- 'ai'는 소문자로 통일됨
- 다만 '시대의', 'python으로', '시작하기'처럼 조사·어미가 단어에 붙어 나오는 한계 확인 → 형태소 분석 필요성 확인

## 실습 7. 모든 제목에서 단어 추출하기

전체 상품명(titles)에 대해 `simple_tokenize()`를 적용해, 단어들을 하나의 리스트(simple_words)로 모은다.

- `for` 반복문으로 제목 하나씩 토큰화한 뒤, 결과를 `simple_words`에 계속 쌓아감
- `append()`는 리스트를 통째로 넣어 리스트 안에 리스트가 생기지만, `extend()`는 단어 하나하나를 풀어서 이어 붙임
- 빈도 분석을 하려면 평평한(단어 하나하나가 나열된) 리스트가 필요하므로 `extend()`를 사용

```python
# 모든 제목에서 단어 추출해서 하나의 리스트로 모으기
simple_words = []
for title in titles:
    words = simple_tokenize(title)
    simple_words.extend(words)

print("전체 단어 수:", len(simple_words))
print(simple_words[:50])

# append()와 extend()의 차이 비교
example1 = []
example1.append(["데이터", "분석"])
print("\nappend() 결과:", example1)

example2 = []
example2.extend(["데이터", "분석"])
print("extend() 결과:", example2)
```

## 실행 결과 확인

199개 제목 전체를 토큰화해서 하나의 리스트(simple_words)로 모았다.

- 전체 단어 수: 760개
- '소년이', '온다', '모순'처럼 조사가 붙은 형태로 단어가 분리됨 (정규표현식 방식의 한계 그대로 나타남)
- append()는 `[['데이터', '분석']]`처럼 리스트 안에 리스트가 생기지만, extend()는 `['데이터', '분석']`처럼 평평하게 이어 붙여져, 빈도 계산에 적합한 형태가 됨을 확인

## 실습 8. Counter로 실제 데이터 빈도 확인하기

토큰화된 simple_words의 상위 30개 단어를 확인하고, 표(DataFrame)로도 정리해본다.

- `Counter(simple_words).most_common(30)`으로 가장 많이 등장한 단어 30개 확인

## 실습 12. Kiwi로 한 문장 품사 확인하기

Kiwi 형태소 분석기로 짧은 문장을 분석해, 단어별 품사(형태소) 정보를 확인한다.

- `kiwi.tokenize(문장)`으로 문장을 형태소 단위로 분리
- 각 토큰의 `form`(형태)과 `tag`(품사)를 확인
- 이번 실습에서 주로 쓸 품사: NNG(일반 명사), NNP(고유 명사), SL(외국어/알파벳)
- 단, 신조어·제품명·인물명·영문 약어·붙여 쓴 표현·브랜드명이 많은 도서 제목 특성상 분석 결과가 예상과 다를 수 있어 반드시 실제 결과를 확인해야 함

```python
from kiwipiepy import Kiwi

kiwi = Kiwi()

sample_title = "AI 시대의 데이터 분석을 위한 파이썬"
tokens = kiwi.tokenize(sample_title)
print(tokens)

print("\n형태와 품사 확인:")
for token in tokens:
    print(token.form, token.tag)
```

## 실행 결과 확인

"AI 시대의 데이터 분석을 위한 파이썬" 문장을 Kiwi로 분석한 결과, 단순 토큰화와 달리 조사와 어미가 깔끔하게 분리됨을 확인했다.

- 명사(NNG/NNP): 시대, 데이터, 분석, 파이썬
- 외국어(SL): AI
- 조사(JKG '의', JKO '을')와 동사/어미(VV '위하', ETM 'ᆫ')는 명사와 분리되어 별도로 추출됨
- 앞서 정규표현식 방식에서 '시대의', '위한'처럼 붙어 나오던 문제가 해결되어, 명사(NNG, NNP)와 외국어(SL)만 골라내면 훨씬 깨끗한 단어 추출이 가능함을 확인

## 실습 13. 명사 중심으로 단어 추출하기

명사(NNG, NNP)와 외국어(SL)만 골라내는 함수를 만들어, 도서 제목에서 의미 있는 단어만 추출한다.

- `TARGET_TAGS`에 원하는 품사(NNG, NNP, SL)를 정해두고, Kiwi 분석 결과에서 이 품사에 해당하는 형태소만 선택
- 선택된 단어는 앞뒤 공백 제거(`strip()`) 후 영문은 소문자로 통일(`lower()`)
- 영문 대소문자를 통일하는 이유: 'AI'와 'ai'를 같은 단어로 집계하기 위함

```python
TARGET_TAGS = {"NNG", "NNP", "SL"}

def extract_words(text):
    result = []
    for token in kiwi.tokenize(str(text)):
        if token.tag in TARGET_TAGS:
            word = token.form.strip().lower()
            result.append(word)
    return result

# 함수 테스트
print(extract_words("AI 시대의 데이터 분석을 위한 파이썬"))
```

## 실습 14. 최소 글자수 필터링 적용하기

한 글자 단어가 너무 많으면 해석이 어려워지므로, 2글자 이상인 단어만 사용하도록 함수를 수정한다.

- `MIN_LENGTH = 2` 기준을 두고, 단어 길이가 이 값 이상일 때만 결과에 포함
- 'ai'처럼 2글자 영문은 그대로 포함됨
- 2글자 기준은 절대적인 규칙이 아니라 이번 분석을 위해 임의로 정한 기준이므로, 보고서 작성 시 "노이즈를 줄이기 위해 2글자 이상의 명사·영문 토큰을 사용했다"처럼 명시하는 것이 좋음

```python
MIN_LENGTH = 2

def extract_words(text):
    result = []
    for token in kiwi.tokenize(str(text)):
        if token.tag in TARGET_TAGS:
            word = token.form.strip().lower()
            if len(word) >= MIN_LENGTH:
                result.append(word)
    return result

# 함수 테스트
print(extract_words("AI 시대의 데이터 분석을 위한 파이썬"))
```

- 테스트 문장에는 원래 1글자 단어가 없어서 결과 변화는 없었지만, 필터 조건(`len(word) >= 2`)이 정상적으로 코드에 반영됨을 확인

## 실습 16. 불용어 이해하기

형태소 분석 후에도 '세트', '개정판', '에디션', '리커버', '한정판'처럼 분석 목적과 관계가 적은 단어가 자주 등장할 수 있어, 이런 단어를 불용어로 지정해 제거하는 방법을 다룬다.

- 불용어를 무조건 많이 제거하는 것은 위험함 → 결과를 보기 전에 임의로 제거하면 원하는 답만 남기는 왜곡이 생길 수 있음
- 권장 순서: 1) 빈도 결과 확인 → 2) 원본 제목에서 실제 쓰임 확인 → 3) 분석 목적과 관련 있는지 판단 → 4) 불용어로 추가 → 5) 결과 다시 확인

## 실습 17. 불용어 목록 만들기

분석 목적과 관계가 적은 단어('세트', '개정판', '에디션', '리커버', '한정판')를 불용어로 지정하고, 이를 제외하도록 `extract_words()` 함수를 수정한다.

- `STOPWORDS`라는 집합(set)에 제외할 단어들을 담아둠
- 함수에 품사 필터(NNG/NNP/SL), 길이 필터(2글자 이상), 불용어 필터(STOPWORDS에 없는 단어) 세 가지 조건을 순서대로 적용
- 처음부터 너무 많은 단어를 넣지 않고, 실제 의미가 있다면 불용어에서 제외해도 됨

```python
TARGET_TAGS = {"NNG", "NNP", "SL"}
MIN_LENGTH = 2
STOPWORDS = {
    "세트",
    "개정판",
    "에디션",
    "리커버",
    "한정판",
}

def extract_words(text):
    result = []
    for token in kiwi.tokenize(str(text)):
        if token.tag not in TARGET_TAGS:
            continue
        word = token.form.strip().lower()
        if len(word) < MIN_LENGTH:
            continue
        if word in STOPWORDS:
            continue
        result.append(word)
    return result

# 불용어가 포함된 실제 제목으로 테스트
test_title = "다크 심리학(20만 부 한정판 스페셜 에디션)"
print(extract_words(test_title))
```

## 실행 결과 확인

불용어 필터가 적용된 `extract_words()`를 실제 제목으로 테스트한 결과, 의도한 대로 동작했다.

- 결과: `['다크', '심리학', '스페셜']`
- STOPWORDS에 등록한 '한정판', '에디션'이 정상적으로 제외됨
- '부', '20만'처럼 길이 필터(2글자 미만)나 품사 필터에 걸리는 표현도 함께 제외되어, 의미 있는 단어만 남음
