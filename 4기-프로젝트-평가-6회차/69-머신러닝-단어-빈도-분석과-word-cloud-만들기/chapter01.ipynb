```python
products = [
    {"name": "키보드", "price":45000, "stock": 12, "category":"전자기기"},
    {"name": "마우스", "price":45000, "stock": 2, "category":"전자기기"},
    {"name": "모니터", "price":45000, "stock": 4, "category":"전자기기"},
    {"name": "책상", "price":45000, "stock": 0, "category":"가구"},
    {"name": "의자", "price":45000, "stock": 7, "category":"가구"},
]

products
```

```python
for_sales=[]

found = False

for product in products:
    if product["stock"] >= 1:
        found = False
        for sale in for_sales:
            if sale["name"] == product["name"]:
                sale["stock"] += product["stock"]
                found = True
                break

        if not found:
            for_sales.append(product.copy())
  
for_sales
```

### 실습 3. 분석에 필요한 컬럼만 선택하기

원본 데이터의 모든 컬럼이 이후 분석에 필요한 것은 아닙니다.

이번 과정에서는 다음 컬럼을 먼저 선택합니다.

```python
# 필요한 컬럼 이름만 리스트로 정리
selected_columns = ['순위', '판매상품 ID', '상품명', '판매가', '인물', '출판사', '발행(출시)일자', '분야']

# df에서 원하는 컬럼만 선택해 새로운 DataFrame df_books 생성
df_books = df[selected_columns]

# 컬럼이 잘 선택되었는지 확인
print("df_books 크기(행, 열):", df_books.shape)   # 컬럼 개수가 8개인지 확인
print("df_books 컬럼:", df_books.columns.tolist())  # 선택한 컬럼 이름이 맞는지 확인

df_books.head()  # 실제 데이터도 한번 확인
```

## 실행 결과 확인

원본 df에서 8개 컬럼만 선택해 `df_books`를 만들었다.

- 크기: (199, 8) — 행 개수는 그대로, 컬럼만 13개 → 8개로 줄어듦
- 선택된 컬럼: 순위, 판매상품 ID, 상품명, 판매가, 인물, 출판사, 발행(출시)일자, 분야
- 이후 분석에 필요한 핵심 컬럼만 남겨 데이터가 한결 보기 편해짐

### 실습 4. 컬럼 이름을 사용하기 쉽게 정리하기 

원본 컬럼 이름은 사람이 읽기에는 좋지만 코드에서 반복해서 사용하기에는 조금 길 수 있습니다.

이번 실습에서는 다음처럼 간단하게 정리할 수 있습니다.

```python
# 바꿀 컬럼 이름을 {기존 이름: 새 이름} 형태의 딕셔너리로 정리
df_books = df_books.rename(columns={
    '판매상품 ID': '판매상품ID',
    '인물': '저자',
    '발행(출시)일자': '발행일'
})

# 컬럼 이름이 잘 바뀌었는지 확인
print(df_books.columns.tolist())
```

## 실행 결과 확인

`rename()`으로 컬럼 이름 3개를 변경했다.

- 판매상품 ID → 판매상품ID
- 인물 → 저자
- 발행(출시)일자 → 발행일
- 나머지 컬럼(순위, 상품명, 판매가, 출판사, 분야)은 그대로 유지됨


### 실습 5. 결측치 확인하기
 

결측치는 값이 비어 있는 데이터를 의미합니다.

 

데이터 분석에서는 결측치를 무조건 삭제하지 않습니다.

 

먼저 어디에 얼마나 있는지 확인하고, 분석 목적에 맞게 처리 방법을 결정합니다.

```python
# 컬럼별 결측치 개수 확인
print(df_books.isna().sum())

# 결측치 개수가 0보다 큰 컬럼만 확인
missing = df_books.isna().sum()
print(missing[missing > 0])
```

## 실행 결과 확인

### 실습 6. 판매가를 숫자로 바꾸기

 

Excel에서 판매가가 다음과 같이 저장되어 있을 수 있습니다.

```python
# 1단계: 쉼표(,)를 빈 문자열로 바꿔서 제거
df_books['판매가'] = df_books['판매가'].str.replace(',', '')

# 2단계: 문자열을 숫자형(정수)으로 변환
df_books['판매가'] = df_books['판매가'].astype(int)

# 3단계: 변환 후 자료형 확인
print(df_books['판매가'].dtype)
```

## 실행 결과 확인

'판매가' 컬럼의 쉼표를 제거하고 숫자형으로 변환했다.

- 변환 전: 문자열(str), 예) '16,200'
- 변환 후: `int64` (정수형)로 확인됨
- 이제 '판매가'로 평균, 합계 등 숫자 계산이 가능해짐

### 실습 7. 발행일을 날짜형으로 바꾸기

 

원본 데이터의 발행일은 다음처럼 보일 수 있습니다.

 

20260701

20260831

20020120

 

이 값은 연도-월-일 정보를 가지고 있지만 바로 날짜 계산에 사용하기 어려울 수 있습니다.

 

날짜 자료형으로 변환하면 연도, 월, 날짜 등을 쉽게 분석할 수 있습니다.

```python
# 1단계: YYYYMMDD 형식의 문자열/숫자를 날짜형으로 변환
df_books['발행일'] = pd.to_datetime(df_books['발행일'], format='%Y%m%d')

# 2단계: 변환 후 자료형 확인
print(df_books['발행일'].dtype)

# 3단계: 앞의 5개 값 확인
print(df_books['발행일'].head())
```

## 실행 결과 확인

'발행일' 컬럼을 `to_datetime()`으로 날짜형으로 변환했다.

- 변환 전: 숫자/문자열, 예) 20140519
- 변환 후: `datetime64[us]` (날짜형)로 확인됨, 예) 2014-05-19
- 이제 연도·월 추출, 기간 비교 등 날짜 관련 연산이 가능해짐

### 실습 8. 문자열 앞뒤 공백 정리하기

 

상품명, 저자, 출판사, 분야는 문자열 데이터입니다.

 

문자열 앞뒤에 불필요한 공백이 있으면 같은 값도 서로 다른 데이터처럼 처리될 수 있습니다.

```python
# 공백을 제거할 컬럼 이름들을 리스트로 정리
text_columns = ['상품명', '저자', '출판사', '분야']

# 반복문으로 각 컬럼에 str.strip() 적용
for col in text_columns:
    df_books[col] = df_books[col].str.strip()

# 처리 후 확인 (컬럼들만 골라서 앞의 5행 확인)
df_books[text_columns].head()
```

## 실행 결과 확인

반복문으로 4개 문자열 컬럼('상품명', '저자', '출판사', '분야')에 `str.strip()`을 적용해 앞뒤 공백을 제거했다.

- 코드 실행 및 결과 출력 정상 확인
- 값 자체는 눈으로 보기에 큰 변화가 없어 보이지만, 보이지 않던 공백이 제거되어 이후 필터링·그룹화 시 오류를 예방할 수 있음

### 실습 9. 중복 데이터 확인하기

 

데이터에서 중복을 발견했다고 해서 바로 삭제하면 안 됩니다.

 

먼저 무엇을 기준으로 중복인지 생각해야 합니다.

 

예를 들어 같은 상품명이 여러 번 등장할 수 있습니다.

 

하지만 같은 제목이라도 다른 판본, 다른 상품, 개정판 등이 있을 수 있습니다.

 

따라서 이번 데이터에서는 상품명만 보고 중복 행을 바로 삭제하지 않습니다.

 

우선 상품을 구분하는 판매상품ID를 확인합니다.

```python
# 1. 전체 행이 완전히 같은 중복 행 개수 확인
dup_all = df_books.duplicated().sum()
print("전체 행 중복 개수:", dup_all)

# 2. 판매상품ID가 중복된 행 개수 확인
dup_id = df_books.duplicated(subset='판매상품ID').sum()
print("판매상품ID 중복 개수:", dup_id)

# 3. 상품명이 중복된 행 개수 확인
dup_name = df_books.duplicated(subset='상품명').sum()
print("상품명 중복 개수:", dup_name)
```

## 실행 결과 확인

전체 행, 판매상품ID, 상품명 기준으로 중복 개수를 각각 확인한 결과 모두 0건으로 나타남

- 중복 데이터가 없어 별도의 중복 제거 작업은 필요하지 않음
